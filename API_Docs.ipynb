# Google Maps Nearby Search API

Below are sample JSON files for a Nearby Search (New) API request and its corresponding response.

## Sample Input JSON

This JSON file represents a Nearby Search (New) request. It specifies that the API should search for places of type "restaurant" within a 500-meter radius of a specified location (latitude 37.7937 and longitude -122.3965). The `maxResultCount` is set to 10, indicating the maximum number of results to return.

```json
{
  "includedTypes": ["restaurant"],
  "maxResultCount": 10,
  "locationRestriction": {
    "circle": {
      "center": {
        "latitude": 37.7937,
        "longitude": -122.3965
      },
      "radius": 500.0
    }
  }
}


````python
import os
import requests
import json

# Set your API key securely (for example, using environment variables)
api_key = os.environ.get('GOOGLE_MAPS_API_KEY')
url = "https://places.googleapis.com/v1/places:searchNearby"

# Define headers with the API key and desired field mask
headers = {
    "Content-Type": "application/json",
    "X-Goog-Api-Key": api_key,
    "X-Goog-FieldMask": "places.displayName,places.formattedAddress,places.types"
}

# Build the JSON payload
payload = {
    "includedTypes": ["restaurant"],
    "maxResultCount": 10,
    "locationRestriction": {
        "circle": {
            "center": {"latitude": 37.7937, "longitude": -122.3965},
            "radius": 500.0
        }
    }
}

# Send the POST request
response = requests.post(url, headers=headers, data=json.dumps(payload))

# Process the response
if response.status_code == 200:
    results = response.json()
    print("Search Results:", results)
else:
    print("Error:", response.status_code, response.text)
````

# Google Maps Routes API Documentation

The Google Maps Routes API allows you to compute a route between an origin and destination. You can call the API via an HTTP POST request (REST) or by using gRPC. This documentation explains both methods with sample code.

---

## Overview

The Routes API provides directions and route information including:
- **Route duration**
- **Distance (in meters)**
- **Polyline (an encoded representation of the route)**

You can specify additional options such as travel mode, routing preferences (e.g., traffic-aware driving), language, units, and route modifiers (e.g., avoid tolls, highways, or ferries).

---

## REST API Usage

### Endpoint

https://routes.googleapis.com/directions/v2:computeRoutes


### How It Works

- **HTTP Method:** POST  
- **Request Body:** JSON containing origin and destination information, travel mode, routing preferences, and other options.
- **Response:** Returns only the fields specified by the field mask provided via the HTTP header `X-Goog-FieldMask` (or `$fields` URL parameter).

### Sample REST Request

The example below computes a driving route with traffic-aware routing, using imperial units and English (US) language. The field mask requests the route duration, distance (in meters), and the encoded polyline.

```bash
curl -X POST -d '{
  "origin": {
    "location": {
      "latLng": {
        "latitude": 37.419734,
        "longitude": -122.0827784
      }
    }
  },
  "destination": {
    "location": {
      "latLng": {
        "latitude": 37.417670,
        "longitude": -122.079595
      }
    }
  },
  "travelMode": "DRIVE",
  "routingPreference": "TRAFFIC_AWARE",
  "computeAlternativeRoutes": false,
  "routeModifiers": {
    "avoidTolls": false,
    "avoidHighways": false,
    "avoidFerries": false
  },
  "languageCode": "en-US",
  "units": "IMPERIAL"
}' \
-H 'Content-Type: application/json' \
-H 'X-Goog-Api-Key: YOUR_API_KEY' \
-H 'X-Goog-FieldMask: routes.duration,routes.distanceMeters,routes.polyline.encodedPolyline' \
'https://routes.googleapis.com/directions/v2:computeRoutes'



````python
import requests
import json

# Define the URL for the computeRoutes endpoint
url = "https://routes.googleapis.com/directions/v2:computeRoutes"

# Define the HTTP headers, including your API key and desired field mask.
headers = {
    "Content-Type": "application/json",
    "X-Goog-Api-Key": "YOUR_API_KEY",  # Replace with your actual API key.
    "X-Goog-FieldMask": "routes.duration,routes.distanceMeters,routes.polyline.encodedPolyline"
}

# Construct the request body with origin, destination, and route options.
body = {
    "origin": {
        "location": {
            "latLng": {
                "latitude": 37.419734,
                "longitude": -122.0827784
            }
        }
    },
    "destination": {
        "location": {
            "latLng": {
                "latitude": 37.417670,
                "longitude": -122.079595
            }
        }
    },
    "travelMode": "DRIVE", 
    "routingPreference": "TRAFFIC_AWARE",
    "computeAlternativeRoutes": False,
    "routeModifiers": {
        "avoidTolls": False,
        "avoidHighways": False,
        "avoidFerries": False
    },
    "languageCode": "en-US",
    "units": "IMPERIAL"
}

# Send the POST request to compute the route.
response = requests.post(url, headers=headers, json=body)

# Check for a successful response.
if response.status_code == 200:
    route_info = response.json()
    print("Route computed successfully:")
    print(json.dumps(route_info, indent=2))
else:
    print("Error computing route:", response.status_code)
    print(response.text)
````

# TripAdvisor Nearby Search API Documentation

The Nearby Search endpoint returns up to 10 locations that are found near a specified latitude and longitude. You can further filter the results using parameters such as category (e.g., hotels, attractions, restaurants, geos), phone number, address, radius, and language.

## Endpoint

GET https://api.content.tripadvisor.com/api/v1/location/nearby_search

## Query Parameters

- **latLong (string, required):**  
  Specifies the latitude and longitude pair around which to search.  
  *Example:* `"39.9042,116.4074"` (for Beijing)

- **key (string, required):**  
  Your Partner API Key for authentication.

- **category (string, optional):**  
  Filter results by type. Valid options include:  
  - `"hotels"`
  - `"attractions"`
  - `"restaurants"`
  - `"geos"`

- **phone (string, optional):**  
  Filter results by phone number. Accepts numbers in any format (omit the "+" sign).

- **address (string, optional):**  
  Filter results by a specific address.

- **radius (string, optional):**  
  Define the search radius from the given latitude/longitude pair.

- **radiusUnit (string, optional):**  
  Unit for the radius. Valid options are `"km"` (kilometers), `"mi"` (miles), or `"m"` (meters).

- **language (string, optional):**  
  Specify the language for the results. Defaults to `"en"` (English).

## HTTP Headers

- **accept:**  
  Example:  
  ```json
  {
    "accept": "application/json"
  }


````python
import requests

# Define your API key and the coordinates for Beijing
api_key = "YOUR_API_KEY"
lat_long = "39.9042,116.4074"  # Coordinates for Beijing
category = "restaurants"       # Filtering for restaurants

# Construct the URL with query parameters
url = (
    "https://api.content.tripadvisor.com/api/v1/location/nearby_search"
    f"?language=en&key={api_key}&latLong={lat_long}&category={category}"
)

# Define the headers to specify the expected response format
headers = {
    "accept": "application/json"
}

# Make the GET request to the API
response = requests.get(url, headers=headers)

# Print the raw JSON response from the API
print(response.text)
````
