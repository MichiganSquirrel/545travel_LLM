```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import pairwise_distances

# Load the data
df = pd.read_csv("../database/Cruise_Preferences_with_UserID.csv")

# Identify feature columns (excluding destination and user_id)
feature_cols = [col for col in df.columns if col not in ["destination", "user_id"]]

# Normalize all feature data
scaler = StandardScaler()
df_scaled = df.copy()
df_scaled[feature_cols] = scaler.fit_transform(df[feature_cols])

# Compute per-user mean feature vectors
user_means = df_scaled.groupby("user_id")[feature_cols].mean()

def recommend_destination(target_user_id):
    """
    For a given user_id, find the nearest neighbor user by K=1 user-based KNN
    on the mean feature vectors, then recommend the destination from that
    neighbor's history whose features are closest to the target user's mean.
    """
    # Extract target user's mean vector
    target_mean = user_means.loc[target_user_id].values.reshape(1, -1)
    
    # Compute distances to all other users' mean vectors
    other_means = user_means.drop(index=target_user_id)
    dists_to_users = pairwise_distances(target_mean, other_means.values).flatten()
    nearest_user_id = other_means.index[np.argmin(dists_to_users)]
    
    # Get the neighbor user's historical records
    neighbor_records = df_scaled[df_scaled["user_id"] == nearest_user_id]
    neighbor_features = neighbor_records[feature_cols].values
    
    # Find which record is closest to the target mean
    dists_to_records = pairwise_distances(target_mean, neighbor_features).flatten()
    best_idx = np.argmin(dists_to_records)
    
    # Return the recommended destination
    return neighbor_records.iloc[best_idx]["destination"]

# Example: recommend for user 1
print(f"Recommended destination for user 1: {recommend_destination(1)}")

# If you want recommendations for all users:
all_recs = {uid: recommend_destination(uid) for uid in user_means.index}
print("\nRecommendations for all users:")
for uid, dest in all_recs.items():
    print(f"User {uid}: {dest}")
```
